# Trial Storage System

## Objective

Create a generic class named Storage[T].

### Requirements

- Store one item of any data type.
- Implement:
  - store()
  - retrieve()
- Demonstrate the class using:
  - Integer
  - String
  - List

```
from typing import Generic, TypeVar

T = TypeVar("T")


class Storage(Generic[T]):

    def __init__(self):
        self.item = None

    def store(self, item: T):
        self.item = item

    def retrieve(self):
        return self.item


number = Storage[int]()
number.store(25)
print("Integer:", number.retrieve())

text = Storage[str]()
text.store("Warehouse")
print("String:", text.retrieve())

items = Storage[list]()
items.store(["Laptop", "Mouse", "Keyboard"])
print("List:", items.retrieve())
```
