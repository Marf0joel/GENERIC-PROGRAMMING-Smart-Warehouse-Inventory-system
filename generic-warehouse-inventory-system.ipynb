# Warehouse Inventory System

## Objective

Create a generic class named Inventory[T].

### Requirements

- Store an item of any type.
- Retrieve the stored item.
- Display the item's type and value.
- Demonstrate using:
  - String
  - Integer
  - Float
  - List
- Extend the class to store multiple items while preserving generic type consistency.

```
from typing import Generic, TypeVar

T = TypeVar("T")


class Inventory(Generic[T]):

    def __init__(self):
        self.items = []

    def store(self, item: T):
        self.items.append(item)

    def retrieve(self):
        return self.items

    def display(self):
        for item in self.items:
            print(f"Type: {type(item).__name__}, Value: {item}")


products = Inventory[str]()
products.store("Laptop")
products.store("Printer")

quantities = Inventory[int]()
quantities.store(50)
quantities.store(120)

prices = Inventory[float]()
prices.store(4500.75)
prices.store(799.99)

serials = Inventory[list]()
serials.store(["SN001", "SN002", "SN003"])
serials.store(["SN004", "SN005"])

print("Products")
products.display()

print()

print("Stock Quantities")
quantities.display()

print()

print("Product Prices")
prices.display()

print()

print("Serial Number Batches")
serials.display()
```
